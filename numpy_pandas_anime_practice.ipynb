```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("anime.csv")
```

```python
df.head(7)
```

```python
def extract_episode(txt):
    data=""
    check=False
    for i in txt:
      if i==")":
          check=False
          return data
      if check==True:
           data+=i
      if i=="(":
          check=True
    
```

```python
df["episode"]=df["Title"].apply(extract_episode)
df
```

```python
df['episode']=df['episode'].str.replace("eps","")
df["episode"]=df["episode"].astype(int)
df
```

```python
def extraction_time(txt):
    
    data=""
    for i in range(len(txt)):
        if txt[i]==")":
            for j in range(i+1,i+20):
               data+=txt[j]
                
            return data
```

```python
df["time"]=df["Title"].apply(extraction_time)
df.head()
```

```python
df[df["episode"]==df["episode"].max()]
```

```python
df.groupby("Title")["episode"].count().sort_values(ascending=False).head
```

```python
df[df["time"]==df["time"].max()]
```
